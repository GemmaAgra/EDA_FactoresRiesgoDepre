INTRODUCCION

Estos datos de 200 personas y se recopilaron para analizar los factores psicológicos y de estilo de vida asociados a la depresión. 
Las variables: estrés, apoyo social, autoestima, horas de sueño por noche, antecedentes familiares de depresión, cafeína, sexo y edad. 
La variable dependiente (depresión) se ha codificado de forma dicotómica (0=No, 1=Sí). 
Objetivo: análisis para identificar factores de riesgo y protección relacionados con la salud mental y predecir su probabilidad en función de variables independientes continuas y categóricas.



Hipótesis 1: El nivel de estrés como factor de riesgo psicológico:

• Enfoque: Variable psicológica.

• Hipótesis de Investigación: A > nivel de estrés, > probabilidad de depresión.
• Hipótesis Nula: El nivel de estrés no tiene un efecto significativo sobre la probabilidad de depresión.


Hipótesis 2: Las horas de sueño como factor protector:

• Enfoque: Variable de estilo de vida.

• Hipótesis de Investigación : Un > nº horas de sueño por noche se asocia con una < probabilidad de depresión.
• Hipótesis Nula: Las horas de sueño no tienen un efecto significativo sobre la probabilidad de depresión.



• Justificación en abmas: El estrés crónico altera los mecanismos biológicos y psicológicos de afrontamiento, uno de los principales predictores de los trastornos del estado de ánimo. Se espera que esta variable tenga un Ratio > 1.


Resultados principales:

• Si el valor p < 0.05, rechazo la Hipótesis Nula y el efecto es significativo.

• Ratio:
	• Para el estrés: Un valor > 1 (ej. 1.35) indica que por cada punto extra de estrés, las probabilidades de depresión aumentan (Hipótesis 1).
	• Para las horas de sueño: Un valor < 1 (ej. 0.78) indica que cada hora extra de sueño disminuye la probabilidad de depresión (Hipótesis 2).
