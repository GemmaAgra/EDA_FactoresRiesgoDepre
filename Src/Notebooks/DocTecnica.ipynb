## Documentación Técnica: EDA:  EXPLORATORY DATA ANALYSIS


Proyecto EDA: análisis de estrés, sueño y depresión


### 1. Descripción General

Factores de riesgo de depresión: 

se recopilaron datos para analizar los factores psicológicos y de estilo de vida asociados a la depresión. 
Variables: estrés, apoyo social, autoestima, horas de sueño por noche, antecedentes familiares de depresión, cafeína, sexo y edad. 
Objetivo: análisis para identificar factores de riesgo de la salud mental y predecir su probabilidad en función de variables categóricas.

Desarrollado en Python usando librerías: NumPy: gestión de matrices, matplotlib.pyplot: visualización de gráficos y pandas.

Se describe la estructura, funcionamiento y requisitos técnicos del programa desarrollado en Python para realizar un análisis exploratorio de datos sobre la relación entre el estrés, las horas de sueño y la depresión.

El programa usa un archivo Excel como fuente de datos y genera dos gráficos de dispersión con sus correspondientes líneas de tendencia y coeficientes de correlación.

class Depresion:

Defino la clase Depresion, cuyo constructor carga los datos.

### 6. Función hipotesis_H1(df)

Objetivo: analizar la relación entre el estrés y la depresión.

Parámetro de entrada: df= DataFrame con los datos del proyecto.

Extracción de vbles/cols del Df: estres = df['estres'] y depre = df['depresion']. 

Cálculo de la correlación:  r = estres.corr(depre). Se calcula el coeficiente de correlación de Pearson entre el estrés y la depresión. (rdo =r).

Cálculo de la línea de tendencia: pendiente, constante = np.polyfit(estres, depre, 1). Se usa np.polyfit() para calcular los parámetros de una recta de 1º grado.

La ecuación obtenida es: y = pendiente * x + constante

pendiente: inclinación de la recta.
constante: punto de corte con el eje Y.

El parámetro 1 indica q se ajusta una recta.

Representación gráfica: creo una figura con plt.figure().

plt.scatter(): representa las observaciones mediante un gráfico de dispersión.
plt.plot(): representa la línea de tendencia.
plt.xlabel(): nombre del eje X.
plt.ylabel(): nombre del eje Y.
plt.title(): título del gráfico.
plt.legend(): leyenda.
plt.grid(): incorpora una cuadrícula.
plt.text(): muestra el coeficiente de correlación.
plt.tight_layout(): ajusta los elementos del gráfico.
plt.show(): muestra el rdo.

El coeficiente de correlación se presenta con 3 decimales con el formato r = {r:.3f}.

### 7. Función hipotesis_H2(df)

Objetivo: analizar la relación entre las horas de sueño y la depresión. 

Extracción de variables: x = df['horas_sueno']  y y = df['depresion']  (horas de sueño es variable independiente y depresión variable dependiente).

Cálculo de la correlación: r = x.corr(y)     Se calcula el coeficiente de correlación de Pearson entre las 2 variables.

Cálculo de la tendencia lineal: pendiente, constante = np.polyfit(x, y, 1)   Se calcula la recta de tendencia q representa la relación lineal entre las horas de sueño y la depresión.

Representación gráfica: Se genera un gráfico de dispersión:

Color verde para los puntos.
Color negro para la línea de tendencia.
Eje X: horas de sueño.
Eje Y: depresión.
Título: relación entre sueño y depresión.
Coeficiente de correlación mostrado en el gráfico.

El gráfico se muestra mediante plt.show().


### 8. Función hipotesis_H3(df)

H3: A > nivel de estrés, < nº de horas de sueño.

Variable independiente (X): estres.

Variable dependiente (Y): horas_sueno.

Relación esperada: correlación negativa (r < 0).

Representación: diagrama de dispersión con línea de tendencia.

Interpretación: se espera que las personas con mayores niveles de estrés duerman menos horas.

### 12. Consideraciones técnicas y posibles errores

Error 1. Archivo Excel no encontrado: si el archivo BD_depresion.xlsx no está en la ubicación esperada.
Sol.: comprobar la estructura de carpetas y el nombre del archivo.

Error 2. Columna no encontrada: Puede aparecer un KeyError si alguna de las columnas no existe o tiene un nombre diferente.
Sol.: comprobar los encabezados del Excel.

Error 3. Librería no instalada: Puede aparecer un ModuleNotFoundError cuando falta alguna dependencia.
Sol.: instalar las librerías indicadas en el apartado de requisitos técnicos.

Error 4. Problemas con valores nulos o datos no numéricos: El cálculo de la correlación puede gestionar determinados valores ausentes, pero np.polyfit() necesita datos numéricos válidos. Comprobar y limpiar los datos antes de ejecutar el análisis.

Error 5. Indentación incorrecta: Python requiere una indentación coherente. Se recomienda usar 4 espacios por nivel y evitar mezclar tabuladores y espacios.
