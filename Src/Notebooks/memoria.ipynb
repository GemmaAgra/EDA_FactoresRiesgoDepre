##  4. Hipótesis de investigación

Hipótesis para estudiar las relaciones entre las variables:

H1: Más estrés → Más probabilidad de depresión.

H2: Más horas de sueño → Menos probabilidad de depresión.

H3: Más estrés → Menos horas de sueño.


### **H1**: El nivel de estrés como factor de riesgo psicológico: relación entre estrés y depresión

Enfoque: Variable psicológica.

Variable independiente (X): Nivel de estrés (E).

Variable dependiente (Y): Probabilidad de depresión (P(D)).

Hipótesis de investigación: Un > nivel de estrés se asocia con una > probabilidad de depresión.

Hipótesis nula: El nivel de estrés no tiene un efecto significativo sobre la probabilidad de depresión.

Representación matemática:

P(D)=a+b⋅E

        P(D): probabilidad de depresión
        E: nivel de estrés
        a: kte.
        b: relación entre estrés y depresión

Relación esperada: Correlación positiva (r>0r>0). b>0 (el efecto del estrés es positivo, a > estrés, > probabilidad de depresión).

Justificación: El estrés crónico puede alterar los mecanismos biológicos y psicológicos de afrontamiento, aumentando la vulnerabilidad emocional y el riesgo de desarrollar trastornos del estado de ánimo.

Representación: Diagrama de dispersión con línea de tendencia ascendente.

Interpretación: Se espera que las personas con > niveles de estrés presenten una > probabilidad de sufrir depresión.


Para estudiar esta hipótesis se usa:

Un gráfico de dispersión q representa el nivel de estrés en el eje X y la vble depresión en el eje Y.
Una línea de tendencia lineal.
El coeficiente de correlación de Pearson.



### **H2**: Las horas de sueño como factor protector: relación entre horas de sueño y depresión

Enfoque: Variable de estilo de vida.

Variable independiente (X): Horas de sueño (S).

Variable dependiente (Y): Probabilidad de depresión (P(D)).

Hipótesis de investigación: Un > nº de horas de sueño se asocia con una < probabilidad de depresión.

Hipótesis nula: Las horas de sueño no tienen un efecto significativo sobre la probabilidad de depresión.

Representación matemática:

P(D)=a−b⋅S

        P(D): probabilidad de depresión
        S: horas de sueño
        a: kte.
        b: relación entre sueño y depresión

Espero: b>0 (se resta el efecto de las horas de sueño, por lo que dormir más reduce la probabilidad de depresión).

Justificación: El sueño adecuado favorece la regulación emocional y los mecanismos de afrontamiento del estrés, actuando como factor protector frente a los trastornos del estado de ánimo.

Relación esperada: correlación negativa (r < 0). (b>0b>0 en la fórmula).

Representación: Diagrama de dispersión con línea de tendencia descendente.

Interpretación: Se espera que las personas que duermen más horas tengan una < probabilidad de sufrir depresión.


Para estudiar esta hipótesis se usa:

Gráfico de dispersión. El nº de horas de sueño se representa en el eje X y la variable depresión en el eje Y.
Línea de tendencia lineal.
Coeficiente de correlación de Pearson.



### **H3**: El nivel de estrés como factor de horas de sueño: A > nivel de estrés, < nº de horas de sueño.

Enfoque: Variable de estilo de vida y bienestar psicológico.

Variable independiente (X): Nivel de estrés (E).

Variable dependiente (Y): Horas de sueño (S).

Hipótesis de investigación: Un > nivel de estrés se asocia con un < nº de horas de sueño.

Hipótesis nula: El nivel de estrés no tiene una relación significativa con el nº de horas de sueño.

Representación matemática:

S=a−b⋅E

        S: horas de sueño
        E: nivel de estrés
        a: kte.
        b: relación entre estrés y horas de sueño

Espero: b>0 (se resta el efecto del estrés, por lo que a mayor estrés, menos horas de sueño).

Justificación: El estrés crónico altera los mecanismos biológicos y psicológicos de afrontamiento, pudiendo dificultar el descanso y reducir las horas de sueño.

Relación esperada: Correlación negativa (r<0r<0).

Representación: Diagrama de dispersión con línea de tendencia descendente.

Interpretación: Se espera que las personas con mayores niveles de estrés duerman menos horas.

Para estudiar esta hipótesis se usa:

Gráfico de dispersión: El nivel de estrés se representa en el eje X y el número de horas de sueño en el eje Y.

Línea de tendencia lineal: Se espera que tenga una pendiente descendente.

Coeficiente de correlación de Pearson: Se espera obtener un valor negativo (r<0r<0), indicando que a > nivel de estrés, < nº de horas de sueño.


### Interpretación esperada de los rdos de las 3 hipótesis:


- (H1) Estrés: un valor > 1 indica q por cada punto extra de estrés, aumentan las probabilidades de depresión.

- (H2) Horas de sueño: un valor < 1 indica q por cada hora extra de sueño, disminuyen las probabilidades de depresión. Las horas de sueño actúan como un factor protector.

- (H3) Estrés: Se espera una correlación negativa (r<0r<0), un valor > 1 indica q por cada punto extra de estrés, disminuye el nº de horas de sueño.
                

En resumen:

H1: Más estrés → Más probabilidad de depresión.

H2: Más horas de sueño → Menos probabilidad de depresión.

H3: Más estrés → Menos horas de sueño.


Nota metodológica:
En H1 y H2 la variable dependiente es la depresión (si está medida como sí/no). En H3 se usa la correlación o la regresión lineal pq la variable dependiente son las horas de sueño.




## 5. Metodología

El proyecto sigue una metodología de análisis exploratorio de datos, dividida en varias fases.

**Fase 1.** Carga de datos: se usa la librería Pandas para importar los datos desde un Excel y almacenarlo en un DataFrame.

**Fase 2.** Exploración inicial: se visualizan las 1ºs filas del conjunto de datos mediante el método head(), para comprobar su contenido y estructura inicial.

**Fase 3.** Análisis estadístico: para cada hipótesis se calcula el coeficiente de correlación de Pearson mediante el método corr() de Pandas.

Este coeficiente permite estudiar la dirección e intensidad de la relación lineal entre las variables.

Su valor oscila entre -1 y 1:


    r próximo a 1: correlación positiva.

    r próximo a -1: correlación negativa.

    r próximo a 0: ausencia de correlación lineal significativa.


**Fase 4.** Representación gráfica: se usan gráficos de dispersión para visualizar las observaciones y líneas de tendencia para representar la dirección general de la relación.

Añado el valor del coeficiente de correlación en cada gráfico para facilitar su interpretación.

##  8. Limitaciones del análisis

Tener en cuenta algunas limitaciones metodológicas:

La correlación no implica causalidad. Una asociación entre 2 variables no demuestra que una sea la causa de la otra.
La variable depresión es binaria, por lo que el coeficiente de Pearson debe interpretarse teniendo en cuenta su naturaleza. En este caso, equivale a una correlación biserial-puntual cuando se compara con una variable cuantitativa.
La línea de tendencia representada es lineal y no constituye un modelo de regresión logística.
El análisis no controla posibles variables externas que puedan influir en los resultados.
No se calculan intervalos de confianza ni pruebas de significación estadística.
