
## Documentación Técnica: EDA:  EXPLORATORY DATA ANALYSIS


### 1. Descripción General

Factores de riesgo de depresión: 

se recopilaron datos para analizar los factores psicológicos y de estilo de vida asociados a la depresión. 
Variables: estrés, apoyo social, autoestima, horas de sueño por noche, antecedentes familiares de depresión, cafeína, sexo y edad. 
Objetivo: análisis para identificar factores de riesgo de la salud mental y predecir su probabilidad en función de variables categóricas.

Desarrollado en Python usando librerías: NumPy: gestión de matrices, matplotlib.pyplot: visualización de gráficos y pandas.


### 2. Arquitectura del Proyecto

El código se divide en 2 bloques:

Lógica de Datos (**utils.py**): Funciones de análisis y representación gráfica.

Fichero principal (**main.py**): Clase q gestiona el inicio del programa y q llamará a las funciones necesarias para ver los gráficos. Matplotlib para visualizar los Rdos.

    

### 3. Especificaciones de los Componentes

**main.py**: carga los datos y controla la ejecución.

Clase Depresion(main.py):   Es el motor principal del programa. Desde aquí llamo a las funciones de utils.py.
                            Se crea un objeto de la clase Depresion.
                            Se ejecuta automáticamente el método __init__().
                            Se importa el Excel BD_depresion.xlsx.        

__init__: Importar, (limpiar y validar datos).

**utils.py**: Funciones de análisis y representación gráfica. Se muestran los participantes analizados y la distribución de depresión.
              Se calculan las correlaciones y se generan los gráficos q representan cómo aumenta la probabilidad de depresión al aumentar el estrés y q a menor horas de sueño mayor 
              probabilidad de depresión.

Eje X: nivel de estrés.

Eje Y: probabilidad de depresión (0-1).

Curva ascendente: a > estrés, > probabilidad de depresión.
