Documentación técnica – Hipótesis H3
1. Nombre de la hipótesis

H3: Existe una relación negativa entre el nivel de estrés y las horas de sueño.

2. Objetivo

Analizar si existe una relación entre el nivel de estrés de las personas y el número de horas que duermen, comprobando si un mayor nivel de estrés está asociado con una menor duración del sueño.

3. Variables usadas
Variable	Tipo	Descripción
estres	Independiente (X)	Nivel de estrés de la persona.
horas_sueno	Dependiente (Y)	Número de horas de sueño.
4. Hipótesis estadística
Hipótesis nula (H0): No existe correlación lineal entre el nivel de estrés y las horas de sueño (ρ = 0).
Hipótesis alternativa (H1): Existe una correlación lineal negativa entre ambas variables (ρ < 0).
5. Metodología

Para estudiar la relación entre las variables se utiliza el coeficiente de correlación de Pearson (r).

Este coeficiente permite medir la intensidad y dirección de la relación lineal entre dos variables cuantitativas.

Su interpretación es:

r próximo a -1: correlación negativa fuerte.
r próximo a 0: ausencia de correlación lineal.
r próximo a +1: correlación positiva fuerte.

Además, se utiliza numpy.polyfit() para calcular una recta de tendencia mediante regresión lineal.

6. Representación gráfica

Se genera un diagrama de dispersión (scatter plot) que representa:

Eje X: nivel de estrés.
Eje Y: horas de sueño.
Puntos azules: observaciones de la base de datos.
Línea negra: tendencia lineal.
Texto informativo: coeficiente de correlación de Pearson.

El gráfico se guarda automáticamente en formato PNG con una resolución de 300 DPI.

Nombre del archivo: H3_estres_sueno.png.

7. Criterio de interpretación

La hipótesis queda respaldada por los datos si se observa una correlación negativa estadísticamente significativa (r < 0 y p < 0,05).

Una correlación negativa indicaría que, a medida que aumenta el nivel de estrés, tienden a disminuir las horas de sueño.

Es importante señalar que una correlación no implica causalidad.
