## TEMA: Factores de riesgo de depresión

Estos datos de 200 personas se recopilaron para analizar los factores psicológicos y de estilo de vida asociados a la depresión. 
Las variables: estrés, apoyo social, autoestima, horas de sueño por noche, antecedentes familiares de depresión, cafeína, sexo y edad. 
La variable dependiente (depresión) se ha codificado de forma dicotómica (0=No, 1=Sí). 
Objetivo: análisis para identificar factores de riesgo de la salud mental y predecir su probabilidad en función de variables categóricas.


#### FUENTES DE DATOS:         Kaggle:  https://www.kaggle.com/datasets/saccatoma/depression-risk-factors-s 
    

## HIPOTESIS

### **H1**: El nivel de estrés como factor de riesgo psicológico:   

• Enfoque: Variable psicológica.

• H. de Investigación: A > nivel de estrés, > probabilidad de depresión.  

P(D)=a+b⋅E   
P(D): probabilidad de depresión
E: nivel de estrés
a: kte
b: relación entre estrés y depresión

Eespero:  b>0


• H. Nula: El nivel de estrés no tiene un efecto significativo sobre la probabilidad de depresión.


### **H2**: Las horas de sueño como factor protector:

• Enfoque: Variable de estilo de vida.

• H. de Investigación : Un > nº horas de sueño se asocia con una < probabilidad de depresión.   

• H. Nula: Las horas de sueño no tienen un efecto significativo sobre la probabilidad de depresión.

P(D)=a−b⋅S
P(D): probabilidad de depresión
S: horas de sueño.
a: kte.
b: relación entre sueño y depresión

espero: b>0 se resta el efecto de las horas de sueño

• Justificación en abmas: El estrés crónico altera los mecanismos biológicos y psicológicos de afrontamiento, uno de los principales predictores de los trastornos del estado de ánimo. (Se espera que esta variable tenga un Ratio > 1).

### **H3**: A > nivel de estrés, < nº de horas de sueño.

Variable independiente (X): estres.

Variable dependiente (Y): horas_sueno.

Relación esperada: correlación negativa (r < 0).

Representación: diagrama de dispersión con línea de tendencia.

Interpretación: se espera que las personas con mayores niveles de estrés duerman menos horas.



### RDOS PPALES:

• Ratio:
	• Estrés: Un valor > 1 indica que por cada punto extra de estrés, las probabilidades de depresión aumentan (H1).
	• Horas de sueño:  Un valor < 1 indica que cada hora extra de sueño disminuye la probabilidad de depresión (H2).
	• Estres: Un valor > 1 indica que por cada punto extra de estrés, se duerme menos horas  (H3).
